```python
# Клітинка 1: Перевірка та імпорт бібліотек
try:
    # Спроба імпорту з новіших версій scipy
    from scipy.optimize import linprog
    print("linprog успішно імпортовано з scipy.optimize")
except ImportError:
    try:
        # Спроба альтернативного імпорту для старих версій
        from scipy.optimize._linprog import linprog
        print("linprog успішно імпортовано з scipy.optimize._linprog")
    except ImportError as e:
        print("Помилка імпорту linprog:", str(e))
        print("Спробуйте оновити scipy: pip install --upgrade scipy")
        raise

import numpy as np
import matplotlib.pyplot as plt

# Додаткова перевірка доступності функції
if 'linprog' not in globals():
    raise ImportError("Функція linprog все ще недоступна. Можливо, проблема з встановленням scipy")
else:
    print("Всі бібліотеки готові до використання")
```

```python
# Клітинка 2: Оптимізація виробництва з перевірками
def production_optimization():
    try:
        print("\n=== Розпочато оптимізацію виробництва ===")
        
        # Дані для оптимізації
        profit = np.array([60, 40])  # Прибуток на одиницю продукції
        constraints = np.array([
            [3, 2],  # Витрати сировини
            [1, 3],  # Витрати часу
            [2, 2]   # Витрати праці
        ])
        resources = np.array([180, 120, 150])  # Доступні ресурси
        
        # Розв'язання задачі лінійного програмування
        result = linprog(
            c=-profit,  # Мінус для максимізації
            A_ub=constraints,
            b_ub=resources,
            bounds=[(0, None), (0, None)],
            method='highs'
        )
        
        if not result.success:
            raise ValueError("Оптимальне рішення не знайдено")
        
        # Вивід результатів
        print(f"\nОптимальний план виробництва:")
        print(f"Продукт A: {result.x[0]:.1f} од.")
        print(f"Продукт B: {result.x[1]:.1f} од.")
        print(f"Максимальний прибуток: {-result.fun:.2f} грн")
        
        # Візуалізація
        plt.figure(figsize=(10, 6))
        
        # Обмеження
        x = np.linspace(0, 80, 100)
        plt.plot(x, (180 - 3*x)/2, label='3A + 2B ≤ 180 (Сировина)')
        plt.plot(x, (120 - x)/3, label='A + 3B ≤ 120 (Час)')
        
        # Область допустимих рішень
        y_min = np.zeros_like(x)
        y_max = np.minimum((180 - 3*x)/2, (120 - x)/3)
        plt.fill_between(x, y_min, y_max, where=(y_max > y_min), alpha=0.2)
        
        # Оптимальна точка
        plt.scatter(result.x[0], result.x[1], color='red', 
                   label=f'Оптимум ({result.x[0]:.1f}, {result.x[1]:.1f})')
        
        # Налаштування графіка
        plt.title('Область допустимих рішень')
        plt.xlabel('Кількість продукту A')
        plt.ylabel('Кількість продукту B')
        plt.legend()
        plt.grid(True)
        plt.xlim(0, 80)
        plt.ylim(0, 100)
        plt.show()
        
    except Exception as e:
        print("\nПомилка під час оптимізації:", str(e))
        print("Можливі причини:")
        print("1. Невірні вхідні дані")
        print("2. Проблеми з бібліотеками")
        print("3. Задача не має розв'язку")
        print("\nСпробуйте:")
        print("1. Перевірити вхідні дані")
        print("2. Оновити бібліотеки: pip install --upgrade scipy numpy matplotlib")
        print("3. Перезапустити ядро (Kernel → Restart)")

# Виклик функції
production_optimization()
```

```python
# Клітинка 3: Приклад 2 - Мінімізація витрат
def minimize_costs():
    print(">>>>> Приклад 2: Мінімізація витрат <<<<<")
    
    # Цільова функція (витрати)
    c = [150, 200]  
    
    # Обмеження (≥ тому мінусуємо)
    A = [[-2, -1],   # Потреба у залізі
         [-1, -2]]   # Потреба у міді
    b = [-100, -150]
    
    # Розв'язання
    res = linprog(c, A_ub=A, b_ub=b, bounds=[(0, None), (0, None)], method='highs')
    
    # Результати
    print(f'Оптимальний видобуток: {res.x[0]:.2f} т з родовища 1, {res.x[1]:.2f} т з родовища 2')
    print(f'Мінімальні витрати: {res.fun:.2f} грн')

minimize_costs()
```

```python
# Клітинка 4: Власна задача
def homework():
    print(">>>>> Власна задача: Оптимізація виробництва <<<<<")
    
    # Вхідні дані
    profit = [60, 40]  # Прибуток за одиницю продукту A і B
    constraints = [
        [3, 2],  # Витрати сировини
        [1, 3],  # Витрати часу
        [2, 2]   # Витрати праці
    ]
    resources = [180, 120, 150]  # Доступні ресурси
    
    # Розв'язання задачі максимізації
    res = linprog(
        c=[-p for p in profit],  # Мінус для максимізації
        A_ub=constraints,
        b_ub=resources,
        bounds=[(0, None), (0, None)],
        method='highs'
    )
    
    # Вивід результатів
    print(f'Оптимальний обсяг виробництва:')
    print(f'Продукт A: {res.x[0]:.2f} од.')
    print(f'Продукт B: {res.x[1]:.2f} од.')
    print(f'Максимальний прибуток: {-res.fun:.2f} грн')
    
    # Візуалізація (перші два обмеження)
    x = np.linspace(0, 80, 400)
    y1 = (180 - 3*x)/2  # 3A + 2B ≤ 180
    y2 = (120 - x)/3    # A + 3B ≤ 120
    
    plt.figure(figsize=(10,6))
    plt.plot(x, y1, label='3A + 2B ≤ 180 (Сировина)')
    plt.plot(x, y2, label='A + 3B ≤ 120 (Час)')
    plt.fill_between(x, np.minimum(y1, y2), 0, alpha=0.2)
    plt.scatter(res.x[0], res.x[1], color='red', label='Оптимальна точка')
    plt.title('Область допустимих рішень')
    plt.xlabel('Продукт A')
    plt.ylabel('Продукт B')
    plt.legend()
    plt.grid()
    plt.xlim(0, 80)
    plt.ylim(0, 100)
    plt.show()

homework()
```
